```python
import numpy as np 
from ctypes import c_int32
q = 8380417
qinv = 58728449
N = 10000
L = 4
```

```python
def i32(n):
    if n < 0:
        n = (1<<32)+n
    return c_int32(n).value

def MR(d):
    g = i32(d*qinv)
    d = (d-g*q)>>32
    return d
```

Load data

```python
u = np.load('attack_10000_u=cs.npy')
c = np.load('attack_10000_c.npy')
s = np.load('attack_s.npy')
```

Verification

```python
c = c.astype(np.int64)
bingo = 1
for i in range(N):
    for j in range(4):
        for k in range(256):
            if MR(c[i][k]*s[j][k]) != u[i][j][k]:
                print(i,j,k)
                bingo = 0
if bingo == 1:
    print('Verification Success!')
```
